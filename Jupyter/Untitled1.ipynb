```python

data = {
    'Item': ['1','1.1','1.1.1','1.1.2','1.1.2.1','1.1.2.1.1','2'],
    'QTY': ['2','3','4','2','2','5','250,00 mm']
    }

data['QTY'] = str(data['QTY'])
data['QTY'] = [int(ele) if ele.isdigit() else ele.rsplit(',', 1)[0] for ele in data['QTY']] 

# Convert string to int
qty = [i for i in data['QTY']]

# Get parent level
ps = [i.rsplit('.', 1)[0] for i in data['Item']]
#print(type(ps[0]))
pp = [i.rpartition('.')[0] for i in data['Item']]
#print(type(pp[0]))

ii = [data['Item'].index(i) for i in ps]
print(type(ii[2]))
print(type(qty))

qt = []
e = 0
for i in pp:
    if i =='':
        qt.append(qty[e])
        #print(i)
    else:
        qt.append(qty[e]*qt[ii[e]])
    #else:
    #    qt.append(ii[e]*qty[e])
    #    #print(test[e])
    e += 1
   
print(qty)
print(qt)
```
